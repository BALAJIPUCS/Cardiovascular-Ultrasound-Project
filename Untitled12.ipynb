```python
# ============================================================
# XAI-Driven Threat Detection and Adversarial Awareness
# in Intelligent CCTV Systems
# ============================================================

# Install Required Packages
!pip -q install ultralytics grad-cam

# ============================================================
# Import Libraries
# ============================================================

import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
from google.colab import files

from ultralytics import YOLO

import torchvision.transforms as transforms
from torchvision import models

from pytorch_grad_cam import GradCAM
from pytorch_grad_cam.utils.image import show_cam_on_image

# ============================================================
# Upload CCTV Image
# ============================================================

print("Upload CCTV Image")

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

# ============================================================
# Read Image
# ============================================================

img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,6))
plt.imshow(img_rgb)
plt.title("Input CCTV Image")
plt.axis("off")
plt.show()

# ============================================================
# Load YOLOv8 Model
# ============================================================

print("Loading YOLOv8 Model...")

yolo_model = YOLO("yolov8n.pt")

# ============================================================
# Threat Detection
# ============================================================

print("\nRunning Threat Detection...")

results = yolo_model(image_path)

annotated_img = results[0].plot()

plt.figure(figsize=(10,8))
plt.imshow(cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB))
plt.title("Threat Detection")
plt.axis("off")
plt.show()

# ============================================================
# Display Detected Objects
# ============================================================

print("\nDetected Objects")

risk_score = 0

for box in results[0].boxes:

    cls = int(box.cls)
    conf = float(box.conf)

    risk_score += conf

    print(
        f"Class ID : {cls} | Confidence : {conf:.3f}"
    )

risk_score = risk_score * 100

# ============================================================
# Threat Level
# ============================================================

if risk_score < 40:
    threat_level = "LOW"

elif risk_score < 70:
    threat_level = "MEDIUM"

else:
    threat_level = "HIGH"

print("\nThreat Risk Score :", round(risk_score,2))
print("Threat Level :", threat_level)

# ============================================================
# Load ResNet50
# ============================================================

print("\nLoading ResNet50...")

resnet = models.resnet50(weights='DEFAULT')

resnet.eval()

# ============================================================
# Preprocess Image
# ============================================================

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

image = Image.open(image_path).convert("RGB")

input_tensor = transform(image)

input_tensor = input_tensor.unsqueeze(0)

# ============================================================
# Original Prediction
# ============================================================

with torch.no_grad():

    original_output = resnet(input_tensor)

original_prediction = torch.argmax(
    original_output,
    dim=1
)

print(
    "\nOriginal Prediction Class:",
    original_prediction.item()
)

# ============================================================
# FGSM Adversarial Attack
# ============================================================

print("\nGenerating FGSM Attack...")

input_adv = input_tensor.clone()

input_adv.requires_grad = True

output = resnet(input_adv)

loss = output.max()

resnet.zero_grad()

loss.backward()

epsilon = 0.05

data_grad = input_adv.grad.data

perturbed_image = (
    input_adv +
    epsilon * data_grad.sign()
)

perturbed_image = torch.clamp(
    perturbed_image,
    0,
    1
)

# ============================================================
# Show Adversarial Image
# ============================================================

adv_img = perturbed_image.squeeze().detach().numpy()

adv_img = np.transpose(
    adv_img,
    (1,2,0)
)

plt.figure(figsize=(6,6))
plt.imshow(adv_img)
plt.title("FGSM Adversarial Image")
plt.axis("off")
plt.show()

# ============================================================
# Prediction After Attack
# ============================================================

with torch.no_grad():

    adv_output = resnet(
        perturbed_image
    )

adv_prediction = torch.argmax(
    adv_output,
    dim=1
)

print(
    "\nAdversarial Prediction Class:",
    adv_prediction.item()
)

# ============================================================
# Adversarial Awareness
# ============================================================

print("\nAdversarial Awareness Check")

if original_prediction.item() != adv_prediction.item():

    print(
        "⚠️ Adversarial Attack Detected"
    )

    attack_status = "Attack Detected"

else:

    print(
        "✅ System Robust"
    )

    attack_status = "No Attack"

# ============================================================
# Grad-CAM Explainability
# ============================================================

print("\nGenerating Grad-CAM...")

target_layer = resnet.layer4[-1]

cam = GradCAM(
    model=resnet,
    target_layers=[target_layer]
)

grayscale_cam = cam(
    input_tensor=input_tensor
)

grayscale_cam = grayscale_cam[0,:]

rgb_img = np.array(
    image.resize((224,224))
)/255.0

visualization = show_cam_on_image(
    rgb_img,
    grayscale_cam,
    use_rgb=True
)

plt.figure(figsize=(8,8))
plt.imshow(visualization)
plt.title("Grad-CAM Explanation")
plt.axis("off")
plt.show()

# ============================================================
# Threat Summary Report
# ============================================================

print("\n")
print("="*60)
print("THREAT ANALYSIS REPORT")
print("="*60)

print("Threat Risk Score :", round(risk_score,2))
print("Threat Level      :", threat_level)
print("Adversarial Check :", attack_status)

print("="*60)

# ============================================================
# Security Alert Engine
# ============================================================

if threat_level == "HIGH":

    print("\n🚨 SECURITY ALERT GENERATED")
    print("Immediate Investigation Recommended")

elif threat_level == "MEDIUM":

    print("\n⚠️ Medium Threat Detected")
    print("Monitoring Recommended")

else:

    print("\n✅ Environment Safe")

# ============================================================
# Save Report
# ============================================================

report = f"""
=================================
XAI CCTV SECURITY REPORT
=================================

Threat Risk Score : {risk_score:.2f}

Threat Level : {threat_level}

Adversarial Status : {attack_status}

=================================
"""

with open(
    "Threat_Report.txt",
    "w"
) as f:

    f.write(report)

print("\nReport Saved : Threat_Report.txt")

# ============================================================
# End of Project
# ============================================================

print("\nProject Execution Completed Successfully")
```

```python
# ==========================================================
# Google Colab Code
# Performance Evaluation for RWMA Detection
# ==========================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from scipy.stats import bootstrap

# ----------------------------------------------------------
# Example data for three coronary territories
# Replace with your model predictions
# ----------------------------------------------------------

np.random.seed(42)

def generate_data(sensitivity, specificity, n=1000):

    positives = n//2
    negatives = n//2

    TP = int(positives*sensitivity)
    FN = positives-TP

    TN = int(negatives*specificity)
    FP = negatives-TN

    y_true = np.concatenate([
        np.ones(positives),
        np.zeros(negatives)
    ])

    y_pred = np.concatenate([
        np.ones(TP),
        np.zeros(FN),
        np.ones(FP),
        np.zeros(TN)
    ])

    scores = y_pred*0.85 + np.random.normal(0,0.10,n)
    scores=np.clip(scores,0,1)

    return y_true,y_pred,scores


territories={

"A":{"sens":0.923,"spec":0.915},
"F":{"sens":0.908,"spec":0.898},
"L":{"sens":0.931,"spec":0.928}

}

results=[]

# ----------------------------------------------------------
# Calculate metrics
# ----------------------------------------------------------

for region,val in territories.items():

    y_true,y_pred,y_score=generate_data(
        val["sens"],
        val["spec"]
    )

    auc=roc_auc_score(y_true,y_score)

    acc=accuracy_score(y_true,y_pred)

    prec=precision_score(y_true,y_pred)

    recall=recall_score(y_true,y_pred)

    f1=f1_score(y_true,y_pred)

    tn,fp,fn,tp=confusion_matrix(
        y_true,
        y_pred
    ).ravel()

    spec=tn/(tn+fp)

    # Bootstrap CI

    rng=np.random.default_rng(42)

    def auc_boot(indices):

        return roc_auc_score(
            y_true[indices],
            y_score[indices]
        )

    idx=np.arange(len(y_true))

    ci=bootstrap(
        (idx,),
        auc_boot,
        confidence_level=0.95,
        n_resamples=1000,
        random_state=42,
        method="percentile"
    )

    results.append([

        region,
        round(auc,3),
        round(recall*100,1),
        round(spec*100,1),
        round(prec*100,1),
        round(f1*100,1),
        f"{ci.confidence_interval.low:.3f}-{ci.confidence_interval.high:.3f}"

    ])


df=pd.DataFrame(

results,

columns=[
"Territory",
"AUROC",
"Sensitivity (%)",
"Specificity (%)",
"Precision (%)",
"F1-score (%)",
"95% CI"
]

)

print(df)

# ----------------------------------------------------------
# Average
# ----------------------------------------------------------

print("\nAverage Performance\n")

print(df.mean(numeric_only=True))
```
